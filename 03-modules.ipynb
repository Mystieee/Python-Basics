## Data serialization with JSON

```python
import json
data ={"name":"mahi", "age":38, "grade":10}

# convert this dict to json
json_str = json.dumps(data)
print(json_str)
print(type(json_str))

# convert tback JSON -> dict
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

## CSV

```python
import csv

#creating a csv file
with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Krish', 12])

# open the cs file and read data
with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# read a file and count number of lines, words and characters
def count_text_file(file_name):
    with open(file_name, 'r') as file:
        lines = file.readlines()
        lines_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        character_count = sum(len(lines) for line in lines)

    return lines_count, word_count, character_count

file_path ='example.txt'
lines, words, chars = count_text_file(file_path)
print(f"lines: {lines} words: {words} characters: {chars}")
```

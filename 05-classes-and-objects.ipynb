```python
## constructor, define attributes of the class
class Dog:
    def __init__(self,name, age): #atrributes , self keyword same like this: accessing the instance var inside the class
        self.name=name #instance variable = parameter object
        self.age=age

#create object
dog=Dog("maxi", 1)
print(dog)
print(dog.name)
```

```python
# define a class with instance methods
class Dog():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # define instance methods
    def bark(self):  #so that we can use instance variables
        print(f"{self.name} Woof!!")


#create obj
dog=Dog("maxi", 0)
dog.bark()

    
```

```python
#Modellling a bank account

class Bank:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def check_balance(self):
        print(f"{self.owner}: balance: {self.balance}")

    def deposit(self, amount):
        self.balance += amount
        print(f"{self.owner}: {amount} is deposited. New balanace is {self.balance}")

    def withdraw(self, amount):
        if self.balance<amount:
            print(f"{self.owner} has insufficient balance to withdraw.")
        else:
            self.balance-= amount
            print(f"{self.owner}: {amount} is withrawed successfully. Available balance is {self.balance}")

o1 =Bank("Person1", 2000)
o2 = Bank("Person2")


o1.check_balance()
o2.check_balance()

o2.withdraw(10)
o2.deposit(100)
o2.withdraw(10)

```

```python
import logging 

#configure basic logging setimgs
logging.basicConfig(level=logging.DEBUG)

#log mesg

logging.debug("this is debug")
logging.warning("wrning")
logging.error("errpr")
logging.info("info")
logging.critical("critical")

# LEVEL: root : message
```

```python
# configuring

import logging 

logging.basicConfig(
    filename='appp.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

logging.debug("this is debug")
logging.warning("wrning")
logging.error("errpr")
logging.info("info")
logging.critical("critical")

#check app.log file

```

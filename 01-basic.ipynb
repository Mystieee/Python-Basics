### list comprehension

List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.


syntax:
Basic syntax: [ expression for item in iterable]

with conditional logic: [expression for item in iterable  if condition]

nested list compre: [[expression] for item1 in itr1   for item2 in itr2]

```python
[x**2 for x in range(10)]
```

```python
## basic list comprehension

# square 
[i**2 for i in range(5)]
```

```python
#find even

[i for i in range(10) if i%2==0]
```

```python
# nested list coprehension -> for 2 lists or momre

n1 = [1,2,3,4]
n2=['a','b','c','d']

[[i,j] for i in n1 for j in n2] 

```

```python
# list comprehension with function calls
list = ["helllllo", "kksosmoisio", "ooiwk", "kjiwujswoisiowwioswoiskw"]

# find length of words
[len(i) for i in list ]
```

```python
# list comprehension with else block

list = [0,1,2,3,4]
['Even' if i%2==0 else 'Odd' for i in list ]
```

```python
## Dict comprehension
# do squares
dict1={x:x**2 for x in range(5)}
print(dict1)
```

```python
# conditional dict comprehension - show only even nos
dict1={x:x**2 for x in range(5) if x%2==0}
print(dict1)
```

```python
# use dict to count freq of elements in list
nos = [1,2,3,3,3,2]
freq={}

for n in nos:
    if n in freq:
        freq[n] +=1
    else:
        freq[n] = 1

print(freq)
```

```python
## • Collection: Inside the function, **kwargs packs all passed key-value pairs into a standard Python dictionary.
#• Keys and Values: The argument names become the dictionary keys, and the values assigned to them become the dictionary values.


#merge 2 dict into 1

d1 = {"a":1, "b":2}
d2= {"b":3, "c":4}

merged_d={**d1, **d2}
print(merged_d)
```

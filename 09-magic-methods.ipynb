```python
## Dunder methods --> double underscore methods __mthodname__


#basic methods:
class Per:
    def __init__(self,name):
        self.name = name

    def __str__(self):
        return f"{self.name} is a good person!"

    def __repr__(self):
        return f"Person(name={self.name})"

p = Per("mah")
print(p) #format str repr
print(repr(p)) #representation



```

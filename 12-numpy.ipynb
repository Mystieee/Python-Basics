```python
''' Shape Shifting:

Attributes: .shape()

Methods: .flatten()  .reshape()
'''



a = np.zeros((3,5))
print(a)

a.shape #-> it specifies the shape
```

```python
# 2D arrays
'''
In 2D arrays:

Rows are --> first dim 

Cols are secon dim

np.flatten() # to flatten the array into just 1 D
'''

a= np.array([[1,2],[2,3],[3,4]])
print(a)
print(f"shape is: {a.shape}")

print("-----")
b= a.flatten() #flatten crates a copy, original remains unchanged!
print(b)
print(f"shape new is: {b.shape}, but the original shape was {a.shape}")


```

```python
''' 

Reshape:

Redefine the shape of the array, without chnging the elements of the array

'''

a=np.array([[1,2],[2,3],[3,4]]) # dim(3,2)

print(a)
print(f"shape of a: {a.shape}")

b= a.reshape((2,3)) #reshpe
print(b)
print(f"reshaped: {b.shape}")

c= a.reshape((3,4)) #we cannot do this 
# print(c)
```

```python
# define the data type:

a = np.array([[2,3],[3,4],[4,5]], dtype=np.int32)
print(a.dtype)
```

```python
# type conversion

# convert boolean to int
a = np.array([[True],[False],[True]], dtype=np.bool)

b= a.astype(np.int32)
print(a)
print(b)
```

```python
# type coercion 

# if array has multiple data types-> it will be converted into single datatype
# as numpy donot support an array to have multple data types...


a= np.array([True, "Boop", 1,2.12])
print(a) # this is converted Coercion

print(a.dtype)
```

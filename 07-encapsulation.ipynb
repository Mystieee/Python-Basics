## Encapsulation:
with Getter or Setter methods

#public private and protected variables

```python
class Person:
    def __init__(self, name, age):
        self.name = name  ## public var
        self.age = age ## public var

    def get_name(person):
        return person.name

    def set_name(person):
        self.person = person.name

p = Person("mahi", 12)
p.get_name()
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name  ## private var
        self.__age = age ## private var
        self.gender = gender  # public var

    def get_name(person):
        return person.__name
    

p = Person("Mhia", 23, "F") 
p.get_name() # tis will throw error, as name is privte and cannot be accessed outside class
```

```python
# protected  single underscore
# you cannot use outside class, but can be accessed from derived class
###

class Person:
    def __init__(self, name, age, gender):
        self._name = name  ## protected var
        self._age = age ## protected var
        self.gender = gender  # public var


class Employee(Person):   # Person is Parent, Employee is child
    def __init__(self, name, age, gender):
        super().__init__( name, age, gender)


em=Employee("Mahi", 12, "F")
print(em._name)
```

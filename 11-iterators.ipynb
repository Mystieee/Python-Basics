```python
# yeild multiple vallues

def my_gen():
    yield 1
    yield 2
    yield 3

gen=my_gen()
gen
```

```python
# another example of generator:
# generator -> creates value on the fly --> using the yield 

#syare nums from 1-n
def sq_num(num):
    for i in range(1, num+1):
        yield i*i


ans = sq_num(3)
print("ths sq:")
print(next(ans))
print(next(ans))
print(next(ans))


'''
yield sends one value at a time without ending the function.

Each next() call resumes the function from where it stopped.

Values are produced only when needed.
'''
```

```python
# reading large files:

def read_file(path):
    with open(path, 'r') as file:
        for line in file:
            yield line

for line in read_file('large.txt'):
    print(line.strip())
```

```python
'''
add extra behaviour to an existing fn

A decorator is a function that takes another function as input and returns a new function.

fn -> input fn -> o/p function


1. Define the decorator first, 
2. then apply it with @decorator_name above the function.


'''

def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

print(myfunction())
```

```python
# function copy


def welcome():
    return "hello welcome"

wel = welcome # function copy, copy one fn to another fn

print(wel()) # copy one fn to another fn


del welcome # delete 

print(wel()) 
```

```python
# closures: fn inside a fn :)
'''
A closure is formed when:

1. A function is defined inside another function (nested function).
2. The inner function references variables from the outer function.
3. The outer function returns the inner function.

How Closures Work Internally?
    When Python creates a closure:

    1. It stores outer function’s variables in a special attribute called __closure__.
    2. The inner function keeps a reference, not a copy, to these variables.

'''
# example 1:
def outer(x):
    def inner(y):
        return x+y; #here x is remembered from outer closures
    return inner

o=outer(10) #x=10
print(o(5)) #y=5 -> call the closures with differne values of y
print(o(9))

o1=outer(100) #again x is remembered!
print(o1(5)) #y=5 -> call the closures with differne values of y
print(o1(9))



```

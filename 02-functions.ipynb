```python
# positional arguments * args

def fn_name(*args):
    for n in args:
        print(n)


fn_name(1,2,3,4,5,"mah")  # here all these will be considered positional args
```

```python
## Key words arguments **kwargs --> all params in {key,value} pairs

def fn_name(**kwargs):
    for k,v in kwargs.items():
        print(f"{k}:{v}")


fn_name(name="mah", age=23, grade="ke")
```

```python
# combine pos args and kwargs

def fn_name (*args, **kwargs):
    for v in args:
        print(f"{v}")

    for k,v in kwargs.items():
        print(f"{k}:{v}")


fn_name(1,2,3, name="mah", age=23, grade="ke")
```

## Lambda Functions

- small, ananymous, un-named functions
- they take any number of args, but return only single expression
------------------------------------------------------------------
- Syntax: lambda arguments : expression
- Example: double = lambda x: x * 2 followed by calling double(5) which returns 10.
- No return Keyword: The result of the single expression is returned automatically.


```python
# lambda args: expr (task or operation)


def add(a,b):
    return a+b


add(2,3)
```

```python
add = lambda a,b: a+b
print(type(add))
print(add(2.3,4))
```

```python
even = lambda n:n%2==0

even(12)
```

```python
# lambda with map -> applies a fn to all items in the list

nos =[1,2,3]

#find square

sq = list(map(lambda x: x*2, nos))
sq
```

## MAps

- The Python map() function applies a specified function to every item in an iterable (like a list or tuple) and returns a lazy iterator (a map object

- map(function, iterable, ...)

```python
n1= [1,2,3]
n2=[4,5,6]

# take each element from the list and add them, like 1+4, 3+5, 3+6
list(map(lambda x,y: x+y, n1,n2))
```

```python
# use map to convert list of string to integers 
str = ["1","2", "3"]
list(map(int, str))
```

```python
# inbuild fn with map

# conevrt to uppercase
f=["apple", "cherry"]
# list(map(str.upper, f)) -> not working

[i.upper() for i in f]
```

```python
def get_name(person):
    return person['name']

people = [
    {"name":"mahi", "age": 12},
    {"name":"kesh", "age":10}
]

list(map(get_name, people))
```

## Filter

- filter out items in a litst or any itr

```python
def even(num):
    if num%2==0:
        return True

num=[1,2,3,4,5,6,7,8]

list(filter(even, num))
```

```python
# filter with lambda function 

# lambda fn --> multiple i/p single output
# lambda expr: fn

num=[1,2,3,4,5,6,7,8] 
print(list(filter(lambda x: x%2==0, num)))
```

```python
# filter with lambda fn and multiple condition
num=[1,2,3,4,5,6,7,8] 

# even and >g

print(list(filter(lambda x: x%2==0 and x>5, num)))
```

```python
# filter on dict

people = [
    {"name":"mahi", "age": 12},
    {"name":"kesh", "age":10}
]

def age_greater_10(person):
    return person['age'] >10

list(filter(age_greater_10, people))
```

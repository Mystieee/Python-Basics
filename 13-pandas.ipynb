### Pandas: Data Frame and Series:

Data structure and analysis tools for diff fns.

Series:
--------
- 1D object: it can hold any data type like: int, float, str


```python
import pandas as pd
```

```python
x= pd.Series([1,2,3,4,5])
x  # index -> default index is created -> len(data)-1
```

```python
# define index for the data:
x = pd.Series([6,7,98,9], index=['a','b','c','d'])
x
```

```python
# look up data using index:
x['c']
```

```python
# we can also use dictionaries for index, value

x={'abc':1, 'def':2, 'xyz':3}
x

a=pd.Series(x)
a
```

```python
# data as a Scalar value: => Data gets repeated for each index defined.

x= pd.Series(3, index=['a','b','c'])
x
```

```python
#merging and joining data frames:
df1 = pd.DataFrame({'Key':['A','B','C'], 'Value1':[1,2,3]})

df2 = pd.DataFrame({'Key':['A','B','D'], 'Value1':[4,5,6]})

#merge two data frames on 'Key column'

pd.merge(df1, df2, on="Key", how="inner") #inner join, based onkey

```

```python
pd.merge(df1, df2, on="Key", how="outer") #outer join, based onkey
```

```python
#left
pd.merge(df1, df2, on="Key", how="left") #left join, based onkey
```

```python
#right
pd.merge(df1, df2, on="Key", how="right") #right join, based onkey
```
